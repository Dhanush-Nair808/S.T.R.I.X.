## 5) Merge person + gun datasets (automated, safe execution with confirmation)
This step will create `/kaggle/working/dataset_merged` with structure expected by YOLO.
Procedure:
- person images/labels will be copied with prefix `person_` and class ids left as 0.
- gun images/labels will be copied with prefix `gun_` and gun class ids remapped from 0 -> 1.
- Finally a merged data.yaml with names ['person','gun'] will be written.
Set MERGE_CONFIRM = True to perform the merge after reviewing the plan output below.

````python
# Configuration for merge
MERGE_CONFIRM = False  # set to True to perform the merge
OUT_ROOT = '/kaggle/working/dataset_merged'

def count_images_labels(root):
    imgs = glob.glob(os.path.join(root, '**', 'images', '*.*'), recursive=True)
    labels = glob.glob(os.path.join(root, '**', 'labels', '*.txt'), recursive=True)
    return len(imgs), len(labels)

print('gun_root:', gun_root)
print('person_root:', person_root)
if os.path.exists(gun_root):
    gi, gl = count_images_labels(gun_root)
    print('Gun dataset images, labels:', gi, gl)
else:
    print('Gun dataset missing; cannot merge')
if person_root and os.path.exists(person_root):
    pi, pl = count_images_labels(person_root)
    print('Person dataset images, labels:', pi, pl)

if not MERGE_CONFIRM:
    print('
MERGE_CONFIRM is False - no changes will be made. Set MERGE_CONFIRM = True to perform the merge')
else:
    import shutil
    # clean output
    if os.path.exists(OUT_ROOT):
        shutil.rmtree(OUT_ROOT)
    for split in ['train','valid','test']:
        os.makedirs(os.path.join(OUT_ROOT, split, 'images'), exist_ok=True)
        os.makedirs(os.path.join(OUT_ROOT, split, 'labels'), exist_ok=True)

    def copy_and_prefix(src_root, prefix, remap_gun_to=None):
        for split in ['train','valid','test']:
            img_src = os.path.join(src_root, split, 'images')
            lbl_src = os.path.join(src_root, split, 'labels')
            if not os.path.exists(img_src):
                continue
            for img_path in glob.glob(os.path.join(img_src, '*.*')):
                fname = os.path.basename(img_path)
                new_name = f'{prefix}_{fname}'
                dst_img = os.path.join(OUT_ROOT, split, 'images', new_name)
                shutil.copy2(img_path, dst_img)
                stem = os.path.splitext(fname)[0]
                lbl_file = os.path.join(lbl_src, f'{stem}.txt')
                if os.path.exists(lbl_file):
                    with open(lbl_file) as f:
                        lines = f.read().splitlines()
                    new_lines = []
                    for L in lines:
                        if not L.strip():
                            continue
                        parts = L.split()
                        cls_id = int(parts[0])
                        if remap_gun_to is not None and cls_id == 0:
                            cls_id = remap_gun_to
                        parts[0] = str(cls_id)
                        new_lines.append(' '.join(parts))
                    dst_lbl = os.path.join(OUT_ROOT, split, 'labels', f'{prefix}_{stem}.txt')
                    with open(dst_lbl, 'w') as outf:
                        outf.write('
'.join(new_lines))

    # copy person dataset (no remap, prefix person_)
    if person_root and os.path.exists(person_root):
        print('Copying person dataset...')
        copy_and_prefix(person_root, 'person', remap_gun_to=None)
    else:
        print('No person dataset provided - merged dataset will contain only gun data mapped to class 1')

    # copy gun dataset and remap gun class 0 -> 1, prefix gun_
    if os.path.exists(gun_root):
        print('Copying gun dataset and remapping gun class 0 -> 1')
        copy_and_prefix(gun_root, 'gun', remap_gun_to=1)
    
    # write merged data.yaml
    merged_yaml = {
        'train': os.path.join(OUT_ROOT, 'train', 'images'),
        'val': os.path.join(OUT_ROOT, 'valid', 'images'),
        'test': os.path.join(OUT_ROOT, 'test', 'images'),
        'nc': 2,
        'names': ['person','gun']
    }
    with open(os.path.join(OUT_ROOT, 'data.yaml'), 'w') as f:
        yaml.dump(merged_yaml, f)

    mi, ml = count_images_labels(OUT_ROOT)
    print('Merged dataset images, labels:', mi, ml)
    print('Merged data.yaml written to', os.path.join(OUT_ROOT, 'data.yaml'))
    print('Now point YOLO to /kaggle/working/dataset_merged/data.yaml for training')
````

## 8) Inference test
Run inference on a sample image using the produced weights and save prediction to /kaggle/working/pred.jpg

````python
from ultralytics import YOLO
import glob, os
sample_img = None
# prefer merged dataset images if exist
if os.path.exists('/kaggle/working/dataset_merged'):
    imgs = glob.glob('/kaggle/working/dataset_merged/**/images/*.*', recursive=True)
else:
    imgs = glob.glob(f'/kaggle/input/{GUN_DATASET_SLUG}/**/images/*.*', recursive=True)
if imgs:
    sample_img = imgs[0]
    print('Running inference on', sample_img)
    weights = '/kaggle/working/person_gun_best.pt' if os.path.exists('/kaggle/working/person_gun_best.pt') else 'yolov8n.pt'
    model = YOLO(weights)
    res = model(sample_img)
    out = '/kaggle/working/pred.jpg'
    try:
        res[0].plot(save=True, fname=out)
        print('Saved prediction to', out)
    except Exception as e:
        print('Inference/plot error:', e)
else:
    print('No sample image found to run inference on')
````
